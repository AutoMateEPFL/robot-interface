```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os
```

```python
cwd = os.getcwd()
parent = os.path.dirname(cwd)
path = os.path.join(parent,'images', 'COMPUTER_VISION_TESTS')
filename_prefix = ['C300_F35_CORNING', 'C340_F35_CORNING', 'C360_F35_CORNING', 'C378_F35_CORNING']
for prefix in filename_prefix :
    original_images = []
    for i in range(7) :
        filename = prefix + f'_0{i}.jpg'
        image_path = os.path.join(path, filename)
        im = cv2.imread(image_path)
        original_images.append((im))

    # fig, ax = plt.subplots(2, 4, figsize=(30,10))
    # for i in range(7) :
    #     x_idx = i if i in [0, 1, 2, 3] else i-4
    #     y_idx = 0 if i in [0, 1,2 , 3] else 1
    #     ax[y_idx][x_idx].imshow(cv2.cvtColor(original_images[i], cv2.COLOR_BGR2RGB))
    # fig.suptitle('Original ' + prefix)
    # plt.show()

    mask_red = []

    lower_red = np.array([160, 190, 0], np.uint8)
    upper_red = np.array([180, 255, 255], np.uint8)

    # Threshold the HSV image to get only red colors
    for i in range(7) :

        mask = cv2.inRange(cv2.cvtColor(original_images[i], cv2.COLOR_BGR2HSV), lower_red, upper_red)
        mask_red.append((mask))

    showing_red_pixels = []
    blue_color = (255, 0, 0) #BGR format

    for i in range(7) :
        result_image = original_images[i].copy()
        result_image[np.where(mask_red[i] != 0)] = blue_color
        showing_red_pixels.append((result_image))

    # fig, ax = plt.subplots(2, 4, figsize=(30,10))
    # for i in range(7) :
    #     x_idx = i if i in [0, 1, 2, 3] else i-4
    #     y_idx = 0 if i in [0, 1,2 , 3] else 1
    #     ax[y_idx][x_idx].imshow(cv2.cvtColor(showing_red_pixels[i], cv2.COLOR_BGR2RGB), cmap='gray')
    # fig.suptitle('Detected red pixels')
    # plt.show()

    x = [0, 1, 2, 3, 4, 5, 6]
    y = []
    for i in range(7) :
        y.append((np.count_nonzero(mask_red[i])))

    plt.figure()
    plt.bar(x, y)
    plt.xlabel('number of plates')
    plt.ylabel('number of detected red pixels ' + prefix)
    plt.show()
```

```python
cwd = os.getcwd()
parent = os.path.dirname(cwd)
path = os.path.join(parent,'images', 'COMPUTER_VISION_TESTS')
C300_F35 = []
for i in range(7) :
    filename = f'C340_F35_CORNING_0{i}.jpg'
    image_path = os.path.join(path, filename)
    im = cv2.imread(image_path)
    C300_F35.append((im))

fig, ax = plt.subplots(2, 4, figsize=(30,10))
for i in range(7) :
    x_idx = i if i in [0, 1, 2, 3] else i-4
    y_idx = 0 if i in [0, 1, 2, 3] else 1
    ax[y_idx][x_idx].imshow(cv2.cvtColor(C300_F35[i], cv2.COLOR_BGR2RGB))
fig.suptitle('Original')
plt.show()

C300_F35_red = []
mask_red = []

lower_red = np.array([160, 190, 0], np.uint8)
upper_red = np.array([180, 255, 255], np.uint8)

# Threshold the HSV image to get only red colors
for i in range(7) :

    mask = cv2.inRange(cv2.cvtColor(C300_F35[i], cv2.COLOR_BGR2HSV), lower_red, upper_red)
    mask_red.append((mask))
    C300_F35_red.append((cv2.bitwise_and(C300_F35[i], C300_F35[i], mask=mask)))

showing_red_pixels = []
blue_color = (255, 0, 0) #BGR format

for i in range(7) :
    result_image = C300_F35[i].copy()
    result_image[np.where(mask_red[i] != 0)] = blue_color
    showing_red_pixels.append((result_image))

fig, ax = plt.subplots(2, 4, figsize=(30,10))
for i in range(7) :
    x_idx = i if i in [0, 1, 2, 3] else i-4
    y_idx = 0 if i in [0, 1,2 , 3] else 1
    ax[y_idx][x_idx].imshow(cv2.cvtColor(showing_red_pixels[i], cv2.COLOR_BGR2RGB), cmap='gray')
fig.suptitle('Detected red pixels')
plt.show()

x = [0, 1, 2, 3, 4, 5, 6]
y = []
for i in range(7) :
    y.append((np.count_nonzero(mask_red[i])))

plt.figure()
plt.bar(x, y)
plt.xlabel('number of plates')
plt.ylabel('number of detected red pixels, C340_F35')
plt.show()
```
